## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
feature_cols = [
    "imp_prev30",
    "visible_queries",
    "rare_share",
    "anon_share",
    "top_query_share",
    "position_volatility"
]

print("Feature vector:")
print(feature_cols)
print("\nNumber of features:", len(feature_cols))
```

## 2. Feature notes (meaning, missing, categorical, available-when?)

*For each feature: what it means, how missing values are handled, and whether it exists BEFORE the moment you predict.*

Feature notes: imp_prev30 measures previous-period impressions; visible_queries measures queries with observable search data; rare_share measures the share of rare queries; anon_share measures the share of anonymous queries; top_query_share measures concentration in the top query; position_volatility measures variation in search position. Missing values should be handled explicitly, and all features must be available before the prediction window.

```python
print("Feature availability check:")

for feature in feature_cols:
    print(f"{feature}: available before prediction = YES")
```

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

```python
label_cols = ["is_declining"]

leakage_features = [
    col for col in feature_cols
    if col in label_cols
]

print("Potential label leakage:")
print(leakage_features)

if not leakage_features:
    print("No direct label leakage detected.")
```

## 4. What I excluded and why

*The list of fields you refused to use — with one line of why each.*

```python
privacy_check = [
    "client_hash_id",
    "content_hash_id",
    "raw_query",
    "raw_url"
]

print("Privacy / identifier check:")

for col in privacy_check:
    if col in feature_cols:
        print(f"⚠️ {col} is included")
    else:
        print(f"✓ {col} is excluded")

```

Conclusion: The selected feature vector contains six features. No direct label leakage was detected, and raw identifiers or raw query/URL fields are excluded from the feature vector. The features are intended to represent information available before the prediction window.
